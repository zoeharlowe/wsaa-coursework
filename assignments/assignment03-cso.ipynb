# assignment03-cso.ipynb

This Jupyter Notebook will retrieve the dataset for the "exchequer account (historical series)" from the CSO, and stores it into a file called "cso.json".

**Author:** Zoe McNamara Harlowe

### Imports

```python
import requests
import json
```

### Source URL

```python
url = "https://ws.cso.ie/public/api.restful/PxStat.Data.Cube_API.ReadDataset/FIQ02/JSON-stat/2.0/en"
```

### Define Function

```python
def getall():
    response = requests.get(url)
    return response.json()
```

### Execute Function

```python
if __name__ == "__main__":
    with open("cso.json", "w") as fp: 
        print(json.dumps(getall(), indent=2), file=fp)
```
